```python
def mostrar_menu():
  print("----------MENU----------")
  print("Escolha uma das opções abaixo")
  print("[1] - Inserir ou Remover Vertice")
  print("[2] - Inserir ou Remover Aresta")
  print("[3] - Calcular e Exibir grau")
  print("[4] - Verificar Aresta")
  print("[5] - Listar Vizinhos")
  print("[6] - Google Maps")
  print("[7] - Exibir Grafo")
  print("[0] - Sair")

def criar_grafo():
    vertices = []
    matriz = []
    return matriz, vertices

def inserir_vertice(matriz, vertices, vertice):
    if vertice not in vertices:
      vertices.append(vertice)

      for linha in matriz:
        linha.append(0)

      nova_linha = [0] * len(vertices)
      matriz.append(nova_linha)
      print(f"{vertice} foi inserido com sucesso!")
    else:
      print("Esse vértice já existe.")
    return matriz, vertices

def inserir_aresta(matriz, vertices, origem, destino, nao_direcionado=False):
    """
    Adiciona uma aresta entre dois vértices.
    """
    if origem not in vertices:
        inserir_vertice(matriz, vertices, origem)
    if destino not in vertices:
        inserir_vertice(matriz, vertices, destino)

    i = vertices.index(origem)
    j = vertices.index(destino)
    matriz[i][j] = 1
    if nao_direcionado:
        matriz[j][i] = 1
    print(f"Aresta de {origem} para {destino} adicionada.")

def remover_vertice(matriz, vertices, vertice):
    if vertice in vertices:
        idx = vertices.index(vertice)
        matriz.pop(idx)
        for linha in matriz:
            linha.pop(idx)
        vertices.remove(vertice)
        print(f"{vertice} foi removido.")
    else:
        print("Vértice não encontrado.")

def remover_aresta(matriz, vertices, origem, destino, nao_direcionado=False):
    if origem in vertices and destino in vertices:
        i = vertices.index(origem)
        j = vertices.index(destino)
        matriz[i][j] = 0
        if nao_direcionado:
            matriz[j][i] = 0
        print("Aresta removida.")
    else:
        print("Origem ou destino não existem.")

def existe_aresta(matriz, vertices, origem, destino):
    if origem in vertices and destino in vertices:
        i = vertices.index(origem)
        j = vertices.index(destino)
        return matriz[i][j] == 1
    return False

def vizinhos(matriz, vertices, vertice):
    if vertice in vertices:
        i = vertices.index(vertice)
        lista = []
        for j in range(len(vertices)):
            if matriz[i][j] == 1:
                lista.append(vertices[j])
        return lista
    return []

def grau_vertices(matriz, vertices):
    graus = {}
    for i, vertice in enumerate(vertices):
        saida = sum(matriz[i])
        entrada = sum(matriz[j][i] for j in range(len(vertices)))
        graus[vertice] = {"saida": saida, "entrada": entrada, "total": saida + entrada}
    return graus

def percurso_valido(matriz, vertices, caminho):
    if len(caminho) < 2:
        return True
    for i in range(len(caminho) - 1):
        if not existe_aresta(matriz, vertices, caminho[i], caminho[i+1]):
            return False
    return True

def listar_vizinhos(matriz, vertices, vertice):
    if vertice in vertices:
        lista = vizinhos(matriz, vertices, vertice)
        print(f"Vizinhos de {vertice}: {lista}")
    else:
        print("Vértice não encontrado.")

def exibir_grafo(matriz, vertices):
    if matriz and vertices:
        print("   " + " ".join(vertices))
        for i, vertice in enumerate(vertices):
            print(f"{vertice}  " + " ".join(str(x) for x in matriz[i]))
    else:
        print("Grafo -> []")

matriz, vertices = criar_grafo()


while True:
  dir_opc = input("Seu grafo é direcionado? [S] - [N]: ")
  dir_opc = dir_opc.upper()
  match dir_opc:
    case 'S':
      dir = True
      break
    case 'N':
      dir = False
      break
    case default:
      print("Direcionamento inválido")

while True:
  try:
    opc = -1
    mostrar_menu()
    print(f"\nMatriz atual: {matriz}")
    print(f"Vertices: {vertices}")
    print(f"Direcionado? {dir}\n")
    opc = int(input())

    match opc:
      case 1:
        using = True
        while using:
          print("Deseja Inserir ou Remover Vertice?")
          print("[I]: Inserir | [R]: Remover")
          print("[S]: Sair")
          opc = input().upper()
          match opc:
            case "I":
              vertice = input("Insira o vertice: ").upper()
              inserir_vertice(matriz, vertices, vertice)
            case "R":
              vertice = input("Insira o vertice a remover: ").upper()
              remover_vertice(matriz, vertices, vertice)
            case "S":
              print("Voltando...")
              using = False
            case _:
              print("Opção inválida")

      case 2:
        using = True
        while using:
          print("Deseja Inserir ou Remover Aresta?")
          print("[I]: Inserir | [R]: Remover")
          print("[S]: Sair")
          opc = input().upper()
          match opc:
            case "I":
              origem = input("Origem: ").upper()
              destino = input("Destino: ").upper()
              inserir_aresta(matriz, vertices, origem, destino, dir)
            case "R":
              origem = input("Origem: ").upper()
              destino = input("Destino: ").upper()
              remover_aresta(matriz, vertices, origem, destino, dir)
            case "S":
              print("Voltando...")
              using = False
            case _:
              print("Opção inválida")
      case 3:
        print(grau_vertices(matriz, vertices))
      case 4:
        origem = input("Origem: ").upper()
        destino = input("Destino: ").upper()
        if existe_aresta(matriz, vertices, origem, destino):
          print("A aresta existe.")
        else:
          print("A aresta não existe.")
      case 5:
        vertice = input("Vértice: ").upper()
        listar_vizinhos(matriz, vertices, vertice)
      case 6:
        caminho = input("Digite o percurso separado por vírgulas: ").upper().replace(" ", "").split(",")
        if percurso_valido(matriz, vertices, caminho):
          print("O percurso é válido!")
        else:
          print("O percurso não é válido!")
      case 7:
        exibir_grafo(matriz, vertices)
      case 0:
        print("Saindo...")
        break
      case default:
        print("Opção inválida")
  except ValueError:
    print("Valor deve ser um numero inteiro")
```

# Atividade 3

```python
def criar_grafo():
    vertices = []
    arestas = []

    return vertices, arestas


def inserir_vertice(vertices, vertice):
    if vertice not in vertices:
        vertices.append(vertice)


def inserir_aresta(vertices, arestas, origem, destino, nao_direcionado=False):
    inserir_vertice(vertices, origem)
    inserir_vertice(vertices, destino)

    arestas.append([origem, destino])

    if nao_direcionado:
        arestas.append([destino, origem])

def remover_aresta(arestas, origem, destino, nao_direcionado=False):
    if [origem, destino] in arestas:
        arestas.remove([origem, destino])

    if nao_direcionado:
        if [destino, origem] in arestas:
            arestas.remove([destino, origem])


def remover_vertice(vertices, arestas, vertice):
    if vertice in vertices:
        vertices.remove(vertice)

        for aresta in arestas[:]:
            if aresta[0] == vertice or aresta[1] == vertice:
                arestas.remove(aresta)


def existe_aresta(arestas, origem, destino):
    for aresta in arestas:
        if aresta == [origem, destino]:
            return True

    return False


def vizinhos(vertices, arestas, vertice):
    lista_vizinhos = []

    for aresta in arestas:
        if aresta[0] == vertice:
            lista_vizinhos.append(aresta[1])

    return lista_vizinhos


def grau_vertices(vertices, arestas):
    graus = {}

    for vertice in vertices:
        graus[vertice] = {"entrada": 0, "saida": 0, "total": 0}

    for aresta in arestas:
        origem = aresta[0]
        destino = aresta[1]

        graus[origem]["saida"] += 1
        graus[destino]["entrada"] += 1

    for vertice in vertices:
        graus[vertice]["total"] = graus[vertice]["entrada"] + graus[vertice]["saida"]

    return graus


def percurso_valido(arestas, caminho):
    for i in range(len(caminho) - 1):
        u = caminho[i]
        v = caminho[i + 1]

        if not existe_aresta(arestas, u, v):
            return False

    return True

def listar_vizinhos(vertices, arestas, vertice):
    if vertice not in vertices:
        print(f"Vértice '{vertice}' não existe no grafo.")
        return

    lista_vizinhos = vizinhos(vertices, arestas, vertice)

    if lista_vizinhos:
        vizinhos_formatados = ", ".join(map(str, lista_vizinhos))
        print(f"Vizinhos de {vertice}: [{vizinhos_formatados}]")
    else:
        print(f"O vértice {vertice} não possui vizinhos de saída.")


def exibir_grafo(vertices, arestas):
    print(f"Vértices: {', '.join(map(str, vertices))}")

    print("Arestas do Grafo:")
    if not arestas:
        print(" O grafo não possui arestas.")
    else:
        for origem, destino in arestas:
            print(f" ({origem} -> {destino})")


def mostrar_menu():
    print("----------MENU----------")
    print("Escolha uma das opções abaixo")
    print("[1] - Inserir ou Remover Vertice")
    print("[2] - Inserir ou Remover Aresta")
    print("[3] - Calcular e Exibir grau")
    print("[4] - Verificar Aresta")
    print("[5] - Listar Vizinhos")
    print("[6] - Verificar Percurso")
    print("[7] - Exibir Grafo")
    print("[0] - Sair")

vertices, arestas = criar_grafo()

while True:
    dir_opc = input("Seu grafo é direcionado? [S] - [N]: ").upper()
    if dir_opc == 'S':
        dir_grafo = True
        break
    elif dir_opc == 'N':
        dir_grafo = False
        break
    else:
        print("Direcionamento inválido")

while True:
    try:
        mostrar_menu()
        print(f"\nArestas atuais: {arestas}")
        print(f"Vertices: {vertices}")
        print(f"Direcionado? {dir_grafo}\n")
        opc = int(input())

        if opc == 1:
            using = True
            while using:
                print("Deseja Inserir ou Remover Vertice?")
                print("[I]: Inserir | [R]: Remover | [S]: Sair")
                sub_opc = input().upper()
                if sub_opc == "I":
                    vertice = input("Insira o vertice: ").upper()
                    inserir_vertice(vertices, vertice)
                    print("Vértice inserido.")
                elif sub_opc == "R":
                    vertice = input("Insira o vertice a remover: ").upper()
                    remover_vertice(vertices, arestas, vertice)
                    print("Vértice removido.")
                elif sub_opc == "S":
                    using = False
                else:
                    print("Opção inválida")
        elif opc == 2:
            using = True
            while using:
                print("Deseja Inserir ou Remover Aresta?")
                print("[I]: Inserir | [R]: Remover | [S]: Sair")
                sub_opc = input().upper()
                if sub_opc == "I":
                    origem = input("Origem: ").upper()
                    destino = input("Destino: ").upper()
                    inserir_aresta(vertices, arestas, origem, destino, not dir_grafo)
                    print("Aresta inserida.")
                elif sub_opc == "R":
                    origem = input("Origem: ").upper()
                    destino = input("Destino: ").upper()
                    remover_aresta(arestas, origem, destino, not dir_grafo)
                    print("Aresta removida.")
                elif sub_opc == "S":
                    using = False
                else:
                    print("Opção inválida")
        elif opc == 3:
            print(grau_vertices(vertices, arestas))
        elif opc == 4:
            origem = input("Origem: ").upper()
            destino = input("Destino: ").upper()
            if existe_aresta(arestas, origem, destino):
                print("A aresta existe.")
            else:
                print("A aresta não existe.")
        elif opc == 5:
            vertice = input("Vértice: ").upper()
            listar_vizinhos(vertices, arestas, vertice)
        elif opc == 6:
            caminho = input("Digite o percurso separado por vírgulas: ").upper().replace(" ", "").split(",")
            if percurso_valido(arestas, caminho):
                print("O percurso é válido!")
            else:
                print("O percurso não é válido!")
        elif opc == 7:
            exibir_grafo(vertices, arestas)
        elif opc == 0:
            print("Saindo...")
            break
        else:
            print("Opção inválida")
    except ValueError:
        print("Valor deve ser um numero inteiro")
    except KeyboardInterrupt:
        print("\nInterrompido pelo usuário.")
        break
```
